## Load and explore the data — 3 marks
Load the local GalaxyMNIST catalogues and image folder. Display six images from each morphology class and report class counts. Select stratified subsets of 4,000 development and 1,000 test images while retaining the supplied split membership. Which classes appear hardest to distinguish visually?

## Prepare data and a baseline — 4 marks
Convert images to 32×32 grayscale, normalize pixels and flatten them. Split development data into 80% fitting and 20% validation, stratified by class. Calculate validation accuracy and macro-F1 for a majority-class predictor based only on fitting labels. Why does flattening not provide rotation invariance?

## Train a classifier — 5 marks
Train a fully connected MLP with hidden layers of 64 and 32 neurons and four output classes. Plot training and validation cross-entropy loss, and validation macro-F1, across epochs. Select a checkpoint using validation macro-F1. What is the difference between the training objective and the checkpoint-selection metric?

## Quantitative and qualitative evaluation — 5 marks
Evaluate the frozen classifier and majority baseline on the test subset. Report accuracy and macro-F1, plot a labelled confusion matrix, and show at least six misclassified examples when available. Which class pairs are confused most often, and what could explain this?

## Test rotation sensitivity — 3 marks
Rotate the test images by 90° and evaluate the same frozen classifier. Report accuracy, macro-F1 and the fraction of predictions that change. Why does similar accuracy before and after rotation not establish prediction invariance? How would you design a separate augmentation study without leaking images across splits?
