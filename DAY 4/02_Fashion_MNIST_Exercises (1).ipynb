# Fashion-MNIST image restoration — Exercise
**Sayan CHAKI · LIRIS · École Centrale de Lyon · Université Lyon 2**

**20 marks · 75–90 minutes**

## Load and explore the images — 3 marks
Load the locally stored Fashion-MNIST image files. Show their shapes and display several original images. Select 3,000 training, 600 validation and 600 test images while preserving the official train/test separation. Why must original images be split before creating degraded copies?

## Create the restoration task — 4 marks
Generate degraded inputs using Gaussian blur with standard deviation 0.8 pixels and additive Gaussian noise with standard deviation 0.12 on intensities normalized to [0,1]. Display original/degraded pairs. What should the regression targets be? Why must the blur act only on spatial axes?

## Compare simple baselines — 3 marks
Measure validation pixel MSE for the degraded input and for a Gaussian smoothing baseline with standard deviation 0.5 pixels. Why is the unchanged input a necessary baseline? What trade-off does smoothing introduce?

## Train a restoration MLP — 5 marks
Flatten images and train a fully connected network with hidden layers of 128, 32 and 128 neurons and 784 output values. Plot training and validation MSE against epoch and select a checkpoint using validation loss. Explain the role and limitations of the bottleneck.

## Evaluate image quality — 5 marks
On held-out test images, compare no processing, smoothing and the MLP using MSE and PSNR. State whether outputs are clipped and report raw MLP MSE as well. Show reference images, degraded inputs, smoothed images, reconstructions and absolute-error maps. Does visually smoother output necessarily have lower error? Finally, evaluate the frozen network at a noise standard deviation of 0.25 and discuss generalization.
