# Laser-spot localization — Exercise
**Sayan CHAKI · LIRIS · École Centrale de Lyon · Université Lyon 2**

**20 marks · 75–90 minutes**

## Load and inspect the data — 3 marks
Load the local COCO annotations and images. How many images contain exactly one annotated laser spot? Why should images with no spot be excluded from a model that always predicts two coordinates? Display several images with their bounding boxes, then compare original and resized images. Is the spot still visible at 64×64 resolution?

## Prepare targets and split the data — 4 marks
How do you convert a box given by left coordinate, top coordinate, width and height into its centre? Normalize the centre by the original image dimensions. Resize RGB images to 64×64, normalize their pixels and flatten them. What are the input and target shapes? Split development data by recording prefix into fitting and validation groups. Why should adjacent frames remain together?

## Establish a baseline — 3 marks
Use the mean spot position in the fitting set as a constant baseline. What validation localization error does it achieve? Why should the baseline use only fitting labels?

## Train an MLP — 5 marks
Train a fully connected regression network with hidden layers of 64 and 32 neurons and two output coordinates. Plot training and validation mean squared error against epoch. Select the checkpoint using validation data. How many trainable parameters does the model have? Do the loss curves suggest overfitting?

## Quantitative and qualitative evaluation — 5 marks
Evaluate the frozen model and baseline on the smartphone test images. Report mean and median pixel distance, mean distance as a percentage of image diagonal, and the percentage of predictions within 20 pixels. Overlay true and predicted centres on at least six original test images. Does the MLP outperform the baseline? Explain two possible causes of failure and one experiment that could investigate them without tuning on test data.
